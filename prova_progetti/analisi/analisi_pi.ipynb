```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.optimize import curve_fit
from matplotlib import colors
```

```python
xp, yp = np.loadtxt("../dati/punti_pi.txt", unpack=True)

xp_acc = []
yp_acc = []

xp_rif = []
yp_rif = []

for i in range(len(xp)):
    distanza = xp[i] ** 2 + yp[i] ** 2

    if distanza < 1.0:
        xp_acc.append(xp[i])
        yp_acc.append(yp[i])
    else:
        xp_rif.append(xp[i])
        yp_rif.append(yp[i])
```

```python
theta = np.linspace(0, np.pi/2, 100)
x=np.cos(theta)
y=np.sin(theta)
```

```python
plt.figure(1)
plt.scatter(xp_acc, yp_acc, None, None,marker='.', color='darkgreen',label='Punti accettati')
plt.scatter(xp_rif, yp_rif, None, None,marker='.', color='navy',label='Punti rifiutati')
plt.plot(x, y, color='Firebrick', linewidth=2.5, linestyle='-',label='Bordo 1/4 di cerchio')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.legend()
plt.title('Simulazione con %d punti' % len(xp))
plt.savefig("grafici/pi.png", bbox_inches="tight") 
plt.show()
```
