```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.optimize import curve_fit
from matplotlib import colors
```

```python
z = np.loadtxt("../dati/dati_rand.txt", unpack=True)
#Definire il vettore che andrà sulle x di modo da plottare la dispersione dei punti campionati
w=list(range(1, len(z) + 1))
```

```python
plt.figure(3)
plt.scatter(w, z, None, None,marker='.', color='navy',label='Valori campionati')
plt.axhline(y=0, color='firebrick', linestyle='-', linewidth=1.5)
plt.xlabel('Estrazione')
plt.ylabel('Valore')
plt.grid(True)
plt.legend()
plt.title('Dispersione estrazione con pdf uniforme tra 0 e 1 con %d punti' % len(z))
plt.savefig("grafici/box_mull.png", bbox_inches="tight") 

plt.show()
```

```python
y = np.loadtxt("../dati/dati_gauss.txt", unpack=True)
#Definire il vettore che andrà sulle x di modo da plottare la dispersione dei punti campionati
x=list(range(1, len(y) + 1))
```

```python
plt.figure(1)
plt.scatter(x, y, None, None,marker='.', color='navy',label='Valori campionati')
plt.axhline(y=0, color='firebrick', linestyle='-', linewidth=1.5)
plt.xlabel('Estrazione')
plt.ylabel('Valore')
plt.grid(True)
plt.legend()
plt.title('Dispersione estrazione con %d punti' % len(y))
plt.savefig("grafici/dispersione_bm.png", bbox_inches="tight") 

plt.show()
```

```python
#Costruire istogramma per vedere la forma gaussiana dei dati campionati
#Genera l'istogramma
#bins=100 divide l'asse orizzontale in 100 intervalli
plt.figure(2)
plt.hist(y, bins=100, color='skyblue', edgecolor='black', alpha=0.7)

# Aggiunge una linea verticale in corrispondenza della media esatta (0)
plt.axvline(x=0, color='firebrick', linestyle='-', linewidth=1.5, label="Media (0.0)")
plt.xlabel('Bin')
plt.ylabel('Valori')
plt.legend()
plt.title(f'Distribuzione dei {len(y)} punti campionati')
plt.savefig("grafici/distribuzione_bm.png", bbox_inches="tight") 

plt.show()
```
